# Mapa 06 - Área de parques por subprefeitura

Esse notebook é responsável pelo download dos dados utilizados no dashboard https://datasp.observatorio.tcm.sp.gov.br/urbanismo#area-de-parques-municipais-e-estaduais-2024 do DataSP. No ano de 2025 os dados foram obtidos manualmente, mas em 2026 serão obtidos via script para os anos de referência de 2024 e 2025.

# Extração dos dados

## Subprefeituras

```python
get_capabilities(filter="subp")
```

```python
subprefeituras = get_features("geoportal:subprefeitura")
subprefeituras
```

## Parques

```python
get_capabilities(filter="parque")
```

```python
parques = get_features("geoportal:GEOSAMPA_cadparcs_parque_unidade_conservacao")
parques
```

# Exportando os arquivos

Neste notebook, vamos apenas salvar os arquivos extraídos na pasta de entrada de dados.

```python
output_dir = path.join('data', 'cache', 'urbanismo')

if not path.exists(output_dir):
    makedirs(output_dir)

for c, gdf in [('subprefeituras', subprefeituras),
               ('parques', parques),
               ]:
    filename=path.join(output_dir, c)
    gdf.to_file(f'{filename}.gpkg', layer=c, driver='GPKG')
```
